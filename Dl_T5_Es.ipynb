## Actividad 1 — Importación de librerías *(sin puntaje, requerida)*

Ejecuta esta celda sin modificaciones.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset

plt.rcParams['figure.figsize'] = (12, 5)
SEED = 42
torch.manual_seed(SEED)
np.random.seed(SEED)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Dispositivo: {device}')
````

---
## Actividad 2 — Preprocesamiento *(2 puntos)*

Completa las líneas marcadas con `# [COMPLETAR]`.
Luego responde la pregunta conceptual en la celda Markdown.

````python
# --- Carga del dataset ---
df = pd.read_csv(DATA_PATH)
print(f'Dimensiones originales: {df.shape}')

# Seleccionar columnas de sensores
sensor_cols = [c for c in df.columns if c.startswith('sensor_')]

# Imputación de nulos: sensor_15 es 100% NaN y algunos sensores tienen huecos
# puntuales. Rellenamos con la media de cada columna (y 0 para columnas sin
# datos) para conservar las 52 features que exige la arquitectura.
df_clean = df[sensor_cols + ['machine_status']].copy()
df_clean[sensor_cols] = df_clean[sensor_cols].fillna(df_clean[sensor_cols].mean())
df_clean[sensor_cols] = df_clean[sensor_cols].fillna(0.0)
df_clean = df_clean.dropna(subset=['machine_status'])

X = df_clean[sensor_cols].values
y = df_clean['machine_status'].values

# [COMPLETAR] Crear un StandardScaler y ajustarlo a X
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# [COMPLETAR] Crear una máscara booleana para los registros de estado NORMAL
mask_normal = (y == 'NORMAL')
mask_anomaly = (y != 'NORMAL')

X_normal  = X_scaled[mask_normal]
X_anomaly = X_scaled[mask_anomaly]

# División 80/20 de datos normales para train/val
n_train   = int(len(X_normal) * 0.8)
X_train   = X_normal[:n_train]
X_val     = X_normal[n_train:]

print(f'Sensores disponibles:          {len(sensor_cols)}')
print(f'Datos de entrenamiento (NORMAL): {len(X_train):,}')
print(f'Datos de validación   (NORMAL):  {len(X_val):,}')
print(f'Datos de anomalía (test):        {len(X_anomaly):,}')
````

 **Pregunta conceptual — Actividad 2:**

¿Por qué se entrena el Autoencoder **solo con datos normales** y no con todo el dataset?
Explica en máximo 3 líneas.

> Se entrena solo con datos NORMAL para que el autoencoder aprenda a reconstruir únicamente el patrón de operación sano. Así, ante una anomalía (BROKEN/RECOVERING) la reconstrucción falla y el error sube, que es justo la señal que usamos para detectarla. Si lo entrenáramos con todo, también aprendería a reconstruir las fallas (borrando la señal) y, además, las anomalías son muy escasas y desbalanceadas.

## Actividad 3 — Definición del Autoencoder *(3 puntos)*

Implementa la clase `AnomalyAutoencoder` con la arquitectura indicada:

```
Encoder: 52 → 32 → 16 → 8   (ReLU en cada capa)
Decoder:  8 → 16 → 32 → 52  (ReLU en capas ocultas, SIN activación en la salida)
```

El método `forward` debe retornar la reconstrucción de la entrada.

````python
class AnomalyAutoencoder(nn.Module):
    def __init__(self, input_dim=52, latent_dim=8):
        super(AnomalyAutoencoder, self).__init__()

        # [COMPLETAR] Definir el encoder: 3 capas lineales con ReLU
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, 32),
            nn.ReLU(),
            nn.Linear(32, 16),
            nn.ReLU(),
            nn.Linear(16, latent_dim),
            nn.ReLU(),
        )

        # [COMPLETAR] Definir el decoder: simétrico, última capa sin activación
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 16),
            nn.ReLU(),
            nn.Linear(16, 32),
            nn.ReLU(),
            nn.Linear(32, input_dim),
        )

    def forward(self, x):
        # [COMPLETAR] pasar x por encoder y luego por decoder
        z = self.encoder(x)
        return self.decoder(z)


# --- Instanciar y verificar ---
INPUT_DIM  = X_train.shape[1]
LATENT_DIM = 8

model = AnomalyAutoencoder(input_dim=INPUT_DIM, latent_dim=LATENT_DIM).to(device)
print(model)

# Prueba rápida de forma
test_input  = torch.randn(4, INPUT_DIM).to(device)
test_output = model(test_input)
print(f'\nEntrada:  {test_input.shape}  →  Salida: {test_output.shape}')
assert test_output.shape == test_input.shape, 'ERROR: las dimensiones de entrada y salida no coinciden.'
print('Verificación de dimensiones: OK')
````

---
## Actividad 4 — Entrenamiento *(2 puntos)*

Completa el bucle de entrenamiento y responde las preguntas conceptuales.

````python
# DataLoader
BATCH_SIZE = 256
train_tensor = torch.FloatTensor(X_train).to(device)
val_tensor   = torch.FloatTensor(X_val).to(device)

train_loader = DataLoader(
    TensorDataset(train_tensor, train_tensor),
    batch_size=BATCH_SIZE, shuffle=True
)

# Configuración
EPOCHS = 50
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

train_losses = []
val_losses   = []

for epoch in range(1, EPOCHS + 1):

    model.train()
    batch_losses = []
    for X_batch, _ in train_loader:
        optimizer.zero_grad()
        output = model(X_batch)                 # [COMPLETAR — forward pass]
        loss   = criterion(output, X_batch)     # [COMPLETAR — calcular MSELoss entre output y X_batch]
        loss.backward()                         # [COMPLETAR — backpropagation y actualización de pesos]
        optimizer.step()
        batch_losses.append(loss.item())

    train_loss = np.mean(batch_losses)
    train_losses.append(train_loss)

    model.eval()
    with torch.no_grad():
        val_out  = model(val_tensor)
        val_loss = criterion(val_out, val_tensor).item()
    val_losses.append(val_loss)

    if epoch % 10 == 0 or epoch == 1:
        print(f'Epoch [{epoch:3d}/{EPOCHS}] | Train Loss: {train_loss:.6f} | Val Loss: {val_loss:.6f}')

# --- Grafica la curva de aprendizaje ---
plt.figure(figsize=(10, 4))
plt.plot(train_losses, label='Entrenamiento', color='#2c5f8a', linewidth=2)
plt.plot(val_losses,   label='Validación',    color='#c0392b', linewidth=2, linestyle='--')
plt.title('Curva de aprendizaje — Autoencoder', fontsize=13, fontweight='bold')
plt.xlabel('Epoch')
plt.ylabel('MSE Loss')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

 **Preguntas conceptuales — Actividad 4:**

**a)** ¿Por qué se usa `MSELoss` como función de pérdida en este problema y no `CrossEntropyLoss`?

> Porque es una tarea de **reconstrucción / regresión** de valores continuos: queremos que la salida del autoencoder se parezca numéricamente a la entrada (las 52 lecturas escaladas de los sensores), y el MSE mide exactamente esa diferencia cuadrática promedio. `CrossEntropyLoss` está pensada para clasificación, donde la salida es una distribución de probabilidad sobre clases discretas, lo cual no aplica aquí porque no estamos prediciendo etiquetas sino reconstruyendo señales.

**b)** Observa la curva de aprendizaje obtenida. ¿El modelo muestra señales de sobreajuste (overfitting)? ¿Cómo lo determinas?

> Sí, hay **señales leves de sobreajuste**. Lo determino comparando ambas curvas: el error de entrenamiento sigue bajando de forma sostenida (de ~0.27 a ~0.075), mientras que el de validación se estanca alrededor de la época 30 (~0.25–0.26) e incluso repunta un poco al final (0.2536 → 0.2609 entre las épocas 40 y 50). Esa brecha amplia y persistente entre train (~0.075) y validación (~0.26), junto con que la validación deja de mejorar mientras el entrenamiento continúa, es el patrón típico de sobreajuste. Aun así es moderado y no impide la detección; podría mitigarse con *early stopping* (~época 30), regularización (weight decay/dropout) o un cuello de botella algo más estrecho.

## Actividad 5 — Detección de anomalías e interpretación *(3 puntos)*

Calcula el error de reconstrucción para datos normales y anómalos, define el umbral y evalúa el modelo.

````python
model.eval()
with torch.no_grad():
    # Error de reconstrucción en validación normal
    val_rec    = model(val_tensor)
    # [COMPLETAR] Calcular el MSE por muestra (media sobre las features, dim=1)
    val_errors = ((val_tensor - val_rec) ** 2).mean(dim=1).cpu().numpy()

    # Error de reconstrucción en datos anómalos
    anom_tensor = torch.FloatTensor(X_anomaly).to(device)
    anom_rec    = model(anom_tensor)
    # [COMPLETAR] Calcular el MSE por muestra para los datos anómalos
    anom_errors = ((anom_tensor - anom_rec) ** 2).mean(dim=1).cpu().numpy()

# [COMPLETAR] Definir el umbral como el percentil 95 de val_errors
threshold = np.percentile(val_errors, 95)

print(f'Error promedio en datos NORMALES:  {val_errors.mean():.6f}')
print(f'Error promedio en datos ANOMALOS:  {anom_errors.mean():.6f}')
print(f'Umbral de deteccion (p95):         {threshold:.6f}')
````

````python
# Visualización: distribución de errores
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(val_errors,  bins=80, alpha=0.7, color='#2c5f8a', label='NORMAL',   density=True)
axes[0].hist(anom_errors, bins=80, alpha=0.7, color='#c0392b', label='ANOMALIA', density=True)
axes[0].axvline(threshold, color='#e67e22', linewidth=2.5, linestyle='--',
                label=f'Umbral p95 = {threshold:.4f}')
axes[0].set_title('Distribución del error de reconstrucción', fontsize=12, fontweight='bold')
axes[0].set_xlabel('MSE de reconstrucción')
axes[0].set_ylabel('Densidad')
axes[0].set_xlim([0, np.percentile(anom_errors, 99)])
axes[0].legend()
axes[0].grid(True, alpha=0.3)

data_box = [val_errors, anom_errors]
bp = axes[1].boxplot(data_box, labels=['NORMAL', 'ANOMALIA'], patch_artist=True)
bp['boxes'][0].set_facecolor('#2c5f8a')
bp['boxes'][1].set_facecolor('#c0392b')
for patch in bp['boxes']:
    patch.set_alpha(0.7)
axes[1].axhline(threshold, color='#e67e22', linewidth=2, linestyle='--',
                label=f'Umbral = {threshold:.4f}')
axes[1].set_title('Comparación de errores por clase', fontsize=12, fontweight='bold')
axes[1].set_ylabel('MSE de reconstrucción')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

````python
# Evaluación: combinar normal y anómalo
all_errors = np.concatenate([val_errors, anom_errors])
y_true = np.concatenate([
    np.zeros(len(val_errors),  dtype=int),
    np.ones(len(anom_errors),  dtype=int)
])
y_pred = (all_errors > threshold).astype(int)

print('=== Reporte de Clasificación ===')
print(classification_report(y_true, y_pred, target_names=['Normal', 'Anomalia']))

# Matriz de confusión
cm = confusion_matrix(y_true, y_pred)
fig, ax = plt.subplots(figsize=(5, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Normal', 'Anomalia'],
            yticklabels=['Normal', 'Anomalia'], ax=ax)
ax.set_title('Matriz de confusión', fontsize=12, fontweight='bold')
ax.set_xlabel('Predicción')
ax.set_ylabel('Valor real')
plt.tight_layout()
plt.show()
````

 **Preguntas conceptuales — Actividad 5:**

**a)** ¿Por qué se usa el **percentil 95** como umbral y no el valor máximo del error de reconstrucción en datos normales?

> El valor máximo es muy sensible a outliers y ruido presentes en los propios datos normales: un único registro normal mal reconstruido dispararía el umbral hacia arriba, dejándolo tan alto que casi ninguna anomalía lo superaría (baja sensibilidad/recall). El percentil 95 es robusto: tolera ese ~5% de normales ruidosos y mantiene el umbral en una zona que sí separa lo normal de lo anómalo.

**b)** Supón que el modelo clasifica como anomalía el **40% de los datos normales de validación**. ¿Qué ajuste harías al umbral y por qué?

> Un 40% de normales marcados como anomalía son demasiados falsos positivos, lo que indica que el umbral está demasiado bajo. Lo **subiría** (por ejemplo, usar el percentil 99 en lugar del 95, o un múltiplo mayor del error normal) para reducir las falsas alarmas, aceptando un pequeño costo en recall. Es un compromiso precisión–recall que se ajusta según cuán caro sea un falso positivo frente a una falla no detectada.

**c)** Menciona **dos ventajas** del Autoencoder para detección de anomalías frente a un clasificador supervisado en el contexto de mantenimiento predictivo minero.

> 1. **No necesita etiquetas de falla:** se entrena solo con datos normales, lo cual es clave en minería donde las fallas reales (BROKEN) son rarísimas y costosas/peligrosas de etiquetar. 2. **Detecta anomalías nuevas o no vistas:** al medir desviación respecto del comportamiento normal, identifica tipos de falla que nunca aparecieron en el entrenamiento, mientras que un clasificador supervisado solo reconoce las clases que vio y sufre con el fuerte desbalance de datos.
